# Application de Kernels de convolution à une image

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

# Affichage de l'image originale

```python
img = cv2.imread("./images/bruce_lee.jpeg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6,6))
plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

# Construction d'une fonction de convolution

```python
def show_filtered_image(img, filtered_img, title, figsize=(6,6)):
    """
    Generic function to display original and filtered image side-by-side.
    Converts BGR to RGB for matplotlib.
    """
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    filtered_rgb = cv2.cvtColor(filtered_img, cv2.COLOR_BGR2RGB)
    
    plt.figure(figsize=figsize)
    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title('Original')
    plt.axis("off")
    
    plt.subplot(1, 2, 2)
    plt.imshow(filtered_rgb)
    plt.title(title)
    plt.axis("off")
    
    plt.tight_layout()
    plt.show()
```

# Box Blur Kernel

```python
box1 = img.copy()
box = cv2.blur(box1, (7, 7))
show_filtered_image(img, box, 'Box Blur (cv2.blur)')
```

# Gaussian Blur Kernel

```python
gauss = cv2.GaussianBlur(img.copy(), (7, 7), 0)
show_filtered_image(img, gauss, 'Gaussian Blur')
```

# Median Blur Kernel

```python
median = cv2.medianBlur(img.copy(), 7)
show_filtered_image(img, median, 'Median Blur')
```

# Bilateral Filter

```python
bilateral = cv2.bilateralFilter(img.copy(), 9, 75, 75)
show_filtered_image(img, bilateral, 'Bilateral Filter')
```

# Sharpening Kernel

```python
kernel = np.array([[0, -1, 0],
                   [-1, 5, -1],
                   [0, -1, 0]])
sharpened = cv2.filter2D(img.copy(), -1, kernel)
show_filtered_image(img, sharpened, 'Sharpening Kernel')
```

# Horizontal Edge Detection Kernel

```python
kernel = np.array([[-1, -1, -1],
                   [0, 0, 0],
                   [1, 1, 1]])
h_edge = cv2.filter2D(img.copy(), -1, kernel)
show_filtered_image(img, h_edge, 'Horizontal Edge Detection Kernel')
```

# Horizontal Sobel Kernel

```python
kernel = np.array([[-1, -2, -1],
                   [0, 0, 0],
                   [1, 2, 1]])
h_sobel = cv2.filter2D(img.copy(), -1, kernel)
show_filtered_image(img, h_sobel, 'Horizontal Sobel Kernel')
```

# Vertical Edge Detection Kernel

```python
kernel = np.array([[-1, 0, 1],
                   [-1, 0, 1],
                   [-1, 0, 1]])
v_edge = cv2.filter2D(img.copy(), -1, kernel)
show_filtered_image(img, v_edge, 'Vertical Edge Detection Kernel')
```
